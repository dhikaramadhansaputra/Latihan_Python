#FOR LOOP

```python
a = [1,2,3,4,5]
```

```python
for i in a:
    print(i)
```

```python
for i in a :
    print(i+1)
```

```python
rasa = ["Vanilla", "Coklat", "Matcha"]
```

```python
for i in rasa:
    print("Rasa",i)
```

```python
topping = ["Kacang", "Misis", "Selai"]
```

```python
for i in rasa:
    for j in topping:
        print("Rasa",i,"Topping",j)
```

```python
cream = {'nama' : 'Dhika',
        'gender' : 'laki-laki',
        'favorit' : ['Coklat', 'Vanilla']}
```

```python
 for i in cream.values():
     print(i)
```

```python
for i,j in cream.items():
    print(i, "---->",j)
```
