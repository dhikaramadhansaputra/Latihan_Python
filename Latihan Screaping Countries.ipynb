#Screaping Screapthissite Countries

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://www.scrapethissite.com/pages/simple/"
page = requests.get(url)
```

```python
print(page.status_code)
```

```python
soup = BeautifulSoup(page.text, 'html')
```

```python
table = soup.find_all('div', class_ = "col-md-4 country")
```

```python
country_list = []
for negara in table:
    raw_country = negara.find_all('h3')
    country_list.extend([x.text.strip() for x in raw_country])
```

```python
country_list
```

```python
capital = []
for raw_capital in table:
    dirty_capital = raw_capital.find_all('span', class_ = 'country-capital')
    capital.extend([x.text.strip() for x in dirty_capital])
```

```python
capital
```

```python
population = []
for raw_population in table:
    dirty_population = raw_population.find_all('span', class_ = 'country-population')
    population.extend([population.text.strip() for population in dirty_population])
```

```python
population
```

```python
country_area = []
for raw_area in table:
    area = raw_area.find_all('span', class_ = 'country-area')
    country_area.extend([x.text.strip() for x in area])
```

```python
country_area
```

Pandas Section

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = ["Negara", "Capital", "Population", "Area"])
```

```python
df['Negara'] = country_list
df['Capital'] = capital
df['Population'] = population
df['Area'] = country_area
```
