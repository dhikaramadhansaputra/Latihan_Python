```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue'

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
table = soup.find_all('table')[0]
```

```python
world_title = table.find_all('th')[1:9]
```

```python
world_table_title = [title.text.strip() for title in world_title]
print(world_table_title)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = world_table_title)
```

```python
column_row = table.find_all('tr')
```

```python
for row in column_row[2:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
    
    length = len(df)
    df.loc[length] = individual_row_data
```
