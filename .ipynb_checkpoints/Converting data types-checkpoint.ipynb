```python
a_int = 7
type(a_int)
```

```python
a_str = "7"
type(a_str)
```

```python
a_convert = int(a_str)
type(a_convert)
```

```python
a_sum = a_int + a_convert
print(a_sum, type(a_convert))
```

```python
a_list = [1,2,3,4,5,2,1]
type(a_list)
```

```python
a_tuple = tuple(a_list)
print(a_tuple, type(a_tuple))
```

```python
a_set = set(a_tuple)
print(a_set, type(a_set))
```
