```python
import os, shutil
```

```python
path = r"D:/Dhika Ramadhan Saputra/Data Analist/Mentahan Automated file sorting/"
```

```python
file_name = os.listdir(path)
```

```python
# Check and Make Folder
folder_name = ["PDF", "CSV", "Microsoft Word", ".Exe", "Journal References", "Image", "SVG", "Power Point", "Videos"]

for folder in range(0, 9):
    if not os.path.exists(path + folder_name[folder]):
        os.makedirs(path + folder_name[folder])
```

```python
#Check and move files into folder
for file in file_name:
    if ".pdf" in file and not os.path.exists(path + "PDF/" + file):
        shutil.move(path + file, path + "PDF/" + file)
    elif ".csv" in file and not os.path.exists(path + "CSV/" + file):
        shutil.move(path + file, path + "CSV/" + file)
    elif ".docx" in file and not os.path.exists(path + "Microsoft Word/" + file):
        shutil.move(path + file, path + "Microsoft Word/" + file)
    elif ".exe" in file and not os.path.exists(path + ".Exe/" + file):
        shutil.move(path + file, path + ".Exe/" + file)
    elif ".ris" in file and not os.path.exists(path + "Journal References/" + file):
        shutil.move(path + file, path + "Journal References/" + file)
    elif ".jpg" or ".png" or ".jpeg" in file and not os.path.exists(path + "Image/" + file):
        shutil.move(path + file, path + "Image/" + file)
    elif ".svg" in file and not os.path.exists(path + "SVG/" + file):
        shutil.move(path + file, path + "SVG/" + file)
    elif ".ppt" or ".pptx" in file and not os.path.exists(path + "Power Point/" + file):
        shutil.move(path + file, path + "Power Point/" + file)
    elif ".mp4" in file and not os.path.exists(path + "Videos/" + file):
        shutil.move(path + file, path + "Videos/" + file)
```
