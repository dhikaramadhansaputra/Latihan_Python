```python
#input
nama = input("Masukan nama anda :")
berat_badan = int(input("Terimakasih "+nama+", Berapa berat badan kamu? (Kg) :"))
tinggi_cm = int(input(str(berat_badan)+" Kg yaa? kalo tinggi kamu berapa? (Cm)"))

#Rumus
tinggi_m = tinggi_cm/100
bmi = berat_badan/(pow(tinggi_m,2))

#Hasil perhitungan
if bmi>0:
    if bmi<=18.5:
        print("Hasil BMI kamu adalah",bmi,"Kamu termasuk KURUS")
    if bmi > 18.5 and bmi < 25:
        print("Hasil BMI kamu adalah",bmi,"Kamu termasuk NORMAL")
    if bmi >= 25:
        print("Hasil BMI kamu adalah",bmi,"Kamu termasuk OBESITAS")
else:
    print("Mohon input data dengan benar")
```
